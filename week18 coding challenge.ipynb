```python
# Task8:Performance Categorization (Data Manipulation)
```

```python
def performance_Category(score):
    if score >= 250:
        return "Excellent"
    elif score >= 200:
        return "Good"
    else:
        return "Needs Improvement"
df['performance_level']= df['total_score'].apply(performance_Category)
df['performance_level'].value_counts()
```
